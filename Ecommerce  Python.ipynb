### 6: Replace Invalid Values


```python
# Replace common invalid values with NaN
# N/A,NULL Replace with NaN

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

### Step 7: Remove Extra Spaces`


```python
df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

### Step 8: Standardize Text Format

```python
# Convert text into Proper Case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

### Step :9 Validate Email Addresses

```python
# Keep only vaild email records.

df = df[df['Email'].str.contains('@',na=False)]
```

### Step 10: Convert Date Columns


```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

### Step 11: Convert Numeric Columns

```python
cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

### Step 12: Remove Invalid Quantity

```python
# Remove records having negative or zero quantity.

df = df[df['Qty']>0]
```

### Step 13: Fill Missing Values

```python
df['Discount'] = df['Discount'].fillna(0)  #"If the discount value is null, replace it with 0."

df['Phone'] = df ['Phone'].fillna ('Unknown') #"If the value is N/A, replace it with 0."

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

### Step 14: Feature Engineering

```python
# Add Total Sales Column.

df['Sales']=df['Qty']*df['Unit_Price'] 
```

```python
# Add Net Amount Columns

df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
# Profit (Example: 20%)
df['Profit']=df['Net_Amount'] # Adding again one columns 
```

```python
df
```

```python
# Order Manth
df['Month'] = df['Order_Date'].dt.month_name() 
# Add Month column. 
```

```python
# Order Year 
df['Year'] = df['Order_Date'].dt.year
# Add Year Column.
```

```python
# Weekday
# Add Week Column.
df['Weekday']=df['Order_Date'].dt.day_name()
```
